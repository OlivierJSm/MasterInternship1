```python
import hictkpy as htk
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.colors import LogNorm, LinearSegmentedColormap

# Import file
f_10k = htk.File("/Users/oliviersmeets/Desktop/University/Master Internship 1/Data Handling/Data/GSE152918_CLP.hic", 10_000)

# Checking characteristics
f_10k.chromosomes()

# Fetch interactions as 
sel = f_10k.fetch("1:10,000,000-20,000,000", join=True)   #Only interactions on chromosome 1
m = sel.to_numpy()

# Define Juicebox color map
juicebox_cmap = LinearSegmentedColormap.from_list(
    "juicebox",
    ["#FFFFFF", "#FFD9A8", "#FF6600", "#990000"],
    N=256
)
juicebox_cmap.set_bad(color="white")

# Define Higlass color map
higlass_cmap = LinearSegmentedColormap.from_list(
    "higlass",
    ["#FFFFFF", "#FFFF00", "#FF4500", "#800080"],
    N=256
)
higlass_cmap.set_bad(color="white")

# Plot
plt.imshow(m, norm=LogNorm(), cmap=juicebox_cmap)
plt.colorbar(label="Interaction counts (log scale)")
plt.title("Hi-C interactions on chr1 (10 kb, 10,000,000-20,000,000)")
plt.show()
```
